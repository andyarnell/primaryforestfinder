```python

import os
import glob

def rename_files(folder_path, substring_to_remove):
    """
    Rename all files in a folder by removing a specific substring from the filename.
    
    Parameters:
    - folder_path: Path to the folder containing files to rename
    - substring_to_remove: Substring to remove from filenames
    """
    # Count for reporting
    renamed_count = 0
    skipped_count = 0
    
    # Get all files in the folder
    files = glob.glob(os.path.join(folder_path, "*"))
    
    print(f"Found {len(files)} files in {folder_path}")
    print(f"Removing substring: '{substring_to_remove}'")
    
    # Process each file
    for file_path in files:
        # Skip directories
        if os.path.isdir(file_path):
            continue
            
        # Get the directory and filename
        directory, filename = os.path.split(file_path)
        
        # Check if the substring exists in the filename
        if substring_to_remove in filename:
            # Create the new filename
            new_filename = filename.replace(substring_to_remove, "")
            new_path = os.path.join(directory, new_filename)
            
            try:
                # Rename the file
                os.rename(file_path, new_path)
                print(f"Renamed: {filename} → {new_filename}")
                renamed_count += 1
            except Exception as e:
                print(f"Error renaming {filename}: {e}")
        else:
            skipped_count += 1
    
    print(f"\nComplete! Renamed {renamed_count} files, skipped {skipped_count} files")
```

```python
# Example usage
if __name__ == "__main__":
    # Replace with your folder path
    folder_path = r"C:\Users\Arnell\OneDrive - Food and Agriculture Organization\project_work\p0002_primary_forest_support\work_in_progress\osm_extracts\osm_regional_250521_geo_csvs"
    
    # Substring to remove (e.g., ".gpkg")
    substring_to_remove = ".gpkg"
    
    rename_files(folder_path, substring_to_remove)
```
